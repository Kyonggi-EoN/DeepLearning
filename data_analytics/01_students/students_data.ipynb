```python
df["Gender"] = df["Gender"].map({"Male":0, "Female":1})
```

```python
df.head()
```

```python
df["Academic_Level"].value_counts()
```

```python
df["Academic_Level"] = df["Academic_Level"].map({"High School":0, "Undergraduate":1, "Graduate":2})
```

```python
df = df.drop(labels="Country", axis=1)
```

```python
df["Most_Used_Platform"].value_counts()
```

```python
platform_map = {
    'Instagram': 'Social_Media',
    'TikTok': 'Social_Media',
    'Facebook': 'Social_Media',
    'Twitter': 'Social_Media',
    'VKontakte': 'Social_Media',
    'WhatsApp': 'Messenger',
    'WeChat': 'Messenger',
    'Snapchat': 'Messenger',
    'LINE': 'Messenger',
    'KakaoTalk': 'Messenger',
    'LinkedIn': 'Others',
    'YouTube': 'Others'
}

df['Platform_Group'] = df['Most_Used_Platform'].map(platform_map).fillna('Others')
```

```python
df["Platform_Group"].value_counts()
```

```python
df = pd.get_dummies(df, columns=["Platform_Group"])
```

```python
df = df.drop("Most_Used_Platform", axis=1)
```

```python
df = df.drop("Relationship_Status", axis=1)
```

```python
df["Affects_Academic_Performance"].value_counts()
```

```python
df["Affects_Academic_Performance"] = df["Affects_Academic_Performance"].map({"No":0, "Yes":1})
```
